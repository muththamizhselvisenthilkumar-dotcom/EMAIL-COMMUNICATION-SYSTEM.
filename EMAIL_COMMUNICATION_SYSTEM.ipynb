```python
# ============================================
# EMAIL COMMUNICATION SYSTEM
# Computer Networks Simulation Application
# SMTP + POP3 Simulation
# ============================================

!pip -q install gradio

import gradio as gr
from datetime import datetime

# --------------------------------------------
# Email Storage
# --------------------------------------------

inbox = []
sent_mail = []

# Sample email
inbox.append({
    "sender": "admin@college.edu",
    "receiver": "student@gmail.com",
    "subject": "Welcome to Computer Networks",
    "message": "Welcome to the Email Communication System.",
    "time": "10:30 AM"
})


# --------------------------------------------
# Send Email
# --------------------------------------------

def send_email(sender, receiver, subject, message):

    if not sender or not receiver or not subject or not message:
        return "❌ Please fill in all the fields."

    current_time = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    email = {
        "sender": sender,
        "receiver": receiver,
        "subject": subject,
        "message": message,
        "time": current_time
    }

    # SMTP simulation
    sent_mail.append(email)

    # Add email to inbox
    inbox.append(email)

    return f"""
✅ EMAIL SENT SUCCESSFULLY

📤 From      : {sender}
📥 To        : {receiver}
📝 Subject   : {subject}
🕐 Time      : {current_time}

Protocol    : SMTP
Status      : Message Delivered Successfully
"""


# --------------------------------------------
# View Inbox
# --------------------------------------------

def view_inbox():

    if not inbox:
        return "📭 Inbox is empty."

    result = "📥 INBOX\n"
    result += "=" * 60 + "\n\n"

    for i, email in enumerate(inbox, 1):

        result += f"""
📧 Email {i}
----------------------------------------
From    : {email['sender']}
To      : {email['receiver']}
Subject : {email['subject']}
Time    : {email['time']}

Message :
{email['message']}

"""

    return result


# --------------------------------------------
# View Sent Mail
# --------------------------------------------

def view_sent():

    if not sent_mail:
        return "📭 No sent emails."

    result = "📤 SENT MAIL\n"
    result += "=" * 60 + "\n\n"

    for i, email in enumerate(sent_mail, 1):

        result += f"""
📧 Email {i}
----------------------------------------
From    : {email['sender']}
To      : {email['receiver']}
Subject : {email['subject']}
Time    : {email['time']}

Message :
{email['message']}

"""

    return result


# ============================================
# APPLICATION INTERFACE
# ============================================

with gr.Blocks(
    title="Email Communication System"
) as app:

    gr.Markdown(
        """
        # 📧 EMAIL COMMUNICATION SYSTEM

        ### Computer Networks Simulation Application

        **SMTP / POP3 - Client / Server Model**
        """
    )

    gr.Markdown(
        """
        This application demonstrates email communication
        between users using SMTP for sending and POP3 for receiving.
        """
    )

    # ----------------------------------------
    # SEND EMAIL
    # ----------------------------------------

    gr.Markdown("## 📤 Compose Email")

    with gr.Row():

        sender = gr.Textbox(
            label="Sender Email",
            placeholder="sender@gmail.com"
        )

        receiver = gr.Textbox(
            label="Receiver Email",
            placeholder="receiver@gmail.com"
        )

    subject = gr.Textbox(
        label="Subject",
        placeholder="Enter email subject"
    )

    message = gr.Textbox(
        label="Message",
        placeholder="Enter your message",
        lines=5
    )

    send_button = gr.Button(
        "📤 Send Email"
    )

    send_output = gr.Textbox(
        label="Email Status",
        lines=9
    )

    send_button.click(
        send_email,
        inputs=[
            sender,
            receiver,
            subject,
            message
        ],
        outputs=send_output
    )

    # ----------------------------------------
    # INBOX
    # ----------------------------------------

    gr.Markdown("## 📥 Inbox")

    inbox_button = gr.Button(
        "🔄 View Inbox"
    )

    inbox_output = gr.Textbox(
        label="Received Emails",
        value=view_inbox(),
        lines=15
    )

    inbox_button.click(
        view_inbox,
        outputs=inbox_output
    )

    # ----------------------------------------
    # SENT MAIL
    # ----------------------------------------

    gr.Markdown("## 📤 Sent Mail")

    sent_button = gr.Button(
        "🔄 View Sent Mail"
    )

    sent_output = gr.Textbox(
        label="Sent Emails",
        lines=15
    )

    sent_button.click(
        view_sent,
        outputs=sent_output
    )


# ============================================
# RUN APPLICATION
# ============================================

app.launch(share=True)
```
